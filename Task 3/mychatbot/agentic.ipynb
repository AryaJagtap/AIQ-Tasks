```python
# 1. Install Google AI SDK & Pandas (Stable & Simple)
!pip install -q -U google-generativeai pandas

import os
import re
import pandas as pd
import google.generativeai as genai
from google.api_core.exceptions import ResourceExhausted


#  PASTE YOUR API KEY HERE
os.environ["GOOGLE_API_KEY"] = "   "

if os.environ["GOOGLE_API_KEY"].startswith("YOUR"):
    print(" STOP: You forgot to paste your API Key!")
else:
    genai.configure(api_key=os.environ["GOOGLE_API_KEY"])

# ==========================================
# 3. DATA PARSING (Using Pandas!)
# ==========================================
def load_data(filename):
    print(f" Parsing {filename} using Pandas...")
    if not os.path.exists(filename):
        print(" Error: File not found. Please upload 'All_Trains_Cleaned.txt'")
        return pd.DataFrame(), {}

    with open(filename, 'r', encoding='utf-8') as f:
        content = f.read()

    # Split by Train Blocks
    blocks = content.split('<TRAIN_START>')
    schedule_rows = []
    train_descriptions = {} # Acts as our "Librarian" knowledge base

    for block in blocks:
        if "TRAIN NO:" not in block: continue

        # Clean up
        clean_block = block.split('<TRAIN_END>')[0].strip()

        # Extract Meta
        t_num = re.search(r"TRAIN NO:\s*(\d+)", clean_block)
        t_name = re.search(r"TRAIN NAME:\s*(.*)", clean_block)

        if t_num:
            no = t_num.group(1)
            name = t_name.group(1).strip() if t_name else "Unknown"

            # Store full text for description search
            train_descriptions[no] = f"Train {no} ({name}). Schedule:\n{clean_block}"

            # Parse Table for Structured Search
            lines = clean_block.split('\n')
            for line in lines:
                if '|' in line and 'STATION NAME' not in line:
                    parts = [p.strip() for p in line.split('|')]
                    if len(parts) >= 4:
                        schedule_rows.append({
                            "train_no": no,
                            "train_name": name,
                            "station": parts[1], # Station Name
                            "arrival": parts[2],
                            "departure": parts[3]
                        })

    # create the DataFrame
    df = pd.DataFrame(schedule_rows)
    print(f" Loaded {len(df)} schedule rows into Pandas.")
    return df, train_descriptions

# Load Data
df, descriptions = load_data("/content/All_Trains_Cleaned.txt")

# ==========================================
# 4. DEFINE TOOLS (The Brains)
# ==========================================

def find_train_route(source: str, destination: str):
    """
    Finds direct trains between a source station and a destination station.
    Args:
        source: The name of the starting station (e.g. 'Thane').
        destination: The name of the ending station (e.g. 'Lasalgaon').
    """
    print(f"\n⚙️ TOOL: Searching Route {source} -> {destination}...")

    # 1. PANDAS FILTERING
    # Find all rows where station matches Source
    src_matches = df[df['station'].str.contains(source, case=False, na=False)]

    # Find all rows where station matches Destination
    dst_matches = df[df['station'].str.contains(destination, case=False, na=False)]

    # 2. LOGIC (Set Intersection)
    # Find train numbers that exist in BOTH lists
    common_train_ids = set(src_matches['train_no']).intersection(set(dst_matches['train_no']))

    if not common_train_ids:
        return f"No direct trains found between {source} and {destination}."

    results = []
    # Loop through the matches to get times
    for t_id in list(common_train_ids)[:10]: # Limit to top 10
        t_name = df[df['train_no'] == t_id].iloc[0]['train_name']

        # Get departure time from Source
        dep_time = src_matches[src_matches['train_no'] == t_id].iloc[0]['departure']

        # Get arrival time at Destination
        arr_time = dst_matches[dst_matches['train_no'] == t_id].iloc[0]['arrival']

        results.append(f"- Train {t_id} ({t_name}): Departs {source} at {dep_time}, Arrives {destination} at {arr_time}")

    return "\n".join(results)

def get_train_details(train_no: str):
    """
    Gets the full schedule and details of a specific train by its number.
    Args:
        train_no: The number of the train (e.g. '11061').
    """
    print(f"\n TOOL: looking up details for {train_no}...")
    return descriptions.get(train_no, "Train details not found in database.")

# ==========================================
# 5. INITIALIZE GEMINI AGENT
# ==========================================
print("\n Initializing Gemini Agent (Native Mode)...")

# Define the tools list
tools_list = [find_train_route, get_train_details]

# Configure the model with tools
model = genai.GenerativeModel(
    model_name='gemini-2.5-flash-lite',
    tools=tools_list,
    system_instruction="""
    You are a Railway Assistant.
    - If the user asks about travel between two points, call 'find_train_route'.
    - If the user asks for details about a specific train number, call 'get_train_details'.
    - Always answer politely based on the tool output.
    """
)

# Start Chat with Automatic Function Calling
chat = model.start_chat(enable_automatic_function_calling=True)

print(" System Ready! (Pandas Powered)")
print("------------------------------------------------")

# ==========================================
# 6. TEST IT
# ==========================================
# ==========================================
# 6. INTERACTIVE CHAT LOOP
# ==========================================
print("\n💬 Chat Session Started! (Type 'exit' or 'quit' to stop)")
print("-------------------------------------------------------")

while True:
    try:
        # 1. Get User Input
        user_input = input("\nYou: ")

        # 2. Check for Exit
        if user_input.lower() in ["exit", "quit", "bye"]:
            print("👋 Exiting chat. Safe travels!")
            break

        # 3. Send to Gemini (It remembers history!)
        response = chat.send_message(user_input)

        # 4. Print Answer
        print(f"AI: {response.text}")

    except Exception as e:
        print(f"⚠️ Error: {e}")
```

```python

!pip install -q -U google-generativeai pandas

import os
import re
import pandas as pd
import google.generativeai as genai
import sys

#  PASTE YOUR API KEY HERE
os.environ["GOOGLE_API_KEY"] = " "

if os.environ["GOOGLE_API_KEY"].startswith("YOUR"):
    print(" STOP: You forgot to paste your API Key in line 14!")
    sys.exit()

genai.configure(api_key=os.environ["GOOGLE_API_KEY"])

# ==========================================
# 2. DATA LOADING (Parses your Text File)
# ==========================================
def load_data(filename):
    print(f" Parsing {filename}...")
    if not os.path.exists(filename):
        print("Error: 'All_Trains_Cleaned.txt' not found. Please upload it!")
        return None, None

    with open(filename, 'r', encoding='utf-8') as f:
        content = f.read()

    # Split by Train Blocks
    blocks = content.split('<TRAIN_START>')
    schedule_rows = []
    train_descriptions = {}

    for block in blocks:
        if "TRAIN NO:" not in block: continue

        # Clean up
        clean_block = block.split('<TRAIN_END>')[0].strip()

        # Extract Meta
        t_num = re.search(r"TRAIN NO:\s*(\d+)", clean_block)
        t_name = re.search(r"TRAIN NAME:\s*(.*)", clean_block)

        if t_num:
            no = t_num.group(1)
            name = t_name.group(1).strip() if t_name else "Unknown"

            # Save full text for "Schedule" lookups
            train_descriptions[no] = f"Train {no} ({name}). Full Schedule:\n{clean_block}"

            # Save rows for "Search" lookups
            lines = clean_block.split('\n')
            for line in lines:
                if '|' in line and 'STATION NAME' not in line:
                    parts = [p.strip() for p in line.split('|')]
                    if len(parts) >= 4:
                        schedule_rows.append({
                            "train_no": no,
                            "train_name": name,
                            "station": parts[1],
                            "arrival": parts[2],
                            "departure": parts[3]
                        })

    df = pd.DataFrame(schedule_rows)
    print(f" Database Ready: {len(df)} rows loaded.")
    return df, train_descriptions

# Run the Loader
df, descriptions = load_data("All_Trains_Cleaned.txt")

if df is None: sys.exit()

# ==========================================
# 3. DEFINE THE 4 UNIVERSAL TOOLS
# ==========================================

def find_route(source: str, destination: str):
    """
    Finds trains running between two specific stations.
    Example: "Thane to Lasalgaon", "How to go from Mumbai to Delhi?"
    """
    print(f"\n TOOL: Connecting {source} -> {destination}...")

    # Filter Source & Destination
    src_df = df[df['station'].str.contains(source, case=False, na=False)]
    dst_df = df[df['station'].str.contains(destination, case=False, na=False)]

    # Find Intersection (Trains that touch both)
    common_ids = set(src_df['train_no']).intersection(set(dst_df['train_no']))

    if not common_ids:
        return f"No direct trains found between {source} and {destination}."

    results = []
    # Loop through common trains to get timings
    for t_id in list(common_ids)[:15]:
        t_name = df[df['train_no'] == t_id].iloc[0]['train_name']
        dep = src_df[src_df['train_no'] == t_id].iloc[0]['departure']
        arr = dst_df[dst_df['train_no'] == t_id].iloc[0]['arrival']
        results.append(f"- {t_id} {t_name} | Dep {source}: {dep} -> Arr {destination}: {arr}")

    return "\n".join(results)

def get_station_trains(station_name: str):
    """
    Lists all trains that stop at a specific station.
    Example: "Trains from Ranchi", "Trains arriving at Ujjain"
    """
    print(f"\n TOOL: Looking up station '{station_name}'...")
    matches = df[df['station'].str.contains(station_name, case=False, na=False)]

    if matches.empty:
        return f"No station found matching '{station_name}'."

    unique_trains = matches[['train_no', 'train_name', 'arrival', 'departure']].drop_duplicates()
    count = len(unique_trains)

    results = [f"Found {count} trains stopping at {station_name}:"]
    for _, row in unique_trains.head(20).iterrows():
        results.append(f"- {row['train_no']} {row['train_name']} (Arr: {row['arrival']} / Dep: {row['departure']})")

    if count > 20:
        results.append(f"...and {count - 20} more.")

    return "\n".join(results)

def search_train_by_name(keyword: str):
    """
    Searches for a train by its name.
    Example: "What is the number of Rajdhani?", "Search for Geetanjali Express"
    """
    print(f"\n TOOL: Searching train name '{keyword}'...")
    matches = df[df['train_name'].str.contains(keyword, case=False, na=False)]
    unique = matches[['train_no', 'train_name']].drop_duplicates().head(10)

    if unique.empty:
        return f"No trains found with name containing '{keyword}'."

    return unique.to_string(index=False)

def get_schedule(train_no: str):
    """
    Get full schedule/route of a specific train number.
    Example: "Route of 12123", "Timetable of train 11061"
    """
    print(f"\n TOOL: Fetching schedule for {train_no}...")
    return descriptions.get(train_no, "Train details not found.")

# ==========================================
# 4. INITIALIZE AGENT
# ==========================================
print("\n Initializing Universal Railway Agent...")

tools_list = [find_route, get_station_trains, search_train_by_name, get_schedule]

model = genai.GenerativeModel(
    model_name='gemini-2.5-flash-lite',
    tools=tools_list,
    system_instruction="""
    You are a Universal Railway Assistant.
    - If user asks A to B? Use 'find_route'.
    - If user asks about one station (from/to/at)? Use 'get_station_trains'.
    - If user searches a name (e.g. Rajdhani)? Use 'search_train_by_name'.
    - If user gives a number? Use 'get_schedule'.
    - Always be helpful and infer the right tool.
    """
)

chat = model.start_chat(enable_automatic_function_calling=True)
print("Agent Ready! (Type 'exit' to quit)")
print("------------------------------------------------")

# ==========================================
# 5. CHAT LOOP
# ==========================================
while True:
    try:
        q = input("\nYou: ")
        if q.lower() in ["exit", "quit", "bye"]:
            print(" Bye!")
            break

        response = chat.send_message(q)
        print(f"AI: {response.text}")
    except Exception as e:
        print(f" Error: {e}")
```
